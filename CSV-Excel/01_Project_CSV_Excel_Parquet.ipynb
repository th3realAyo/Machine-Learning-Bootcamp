## Load the Data

```python
df_csv = pd.read_csv('titanic.csv')
df_csv.head()
```

```python
rows, columns = df_csv.shape
print(f"Rows: {rows} : Columns: {columns}")

# or 

print(f"Rows: {df_csv.shape[0]}, Columns: {df_csv.shape[1]}")
```

```python
# Load selected columns

df_csv_subset = pd.read_csv(
    filepath_or_buffer = 'titanic.csv',
    usecols = ['Survived', 'Pclass', 'Age', 'Fare']
)

df_csv_subset.head()
```

```python
# Load N number of rows
df_csv_rows = pd.read_csv("titanic.csv", nrows=5)

rows, columns = df_csv_rows.shape
print(f"Rows: {rows} : Columns: {columns}")
```

**Load Excel File**

```python
df_excel = pd.read_excel("titanic.xlsx")

df_excel.head()
```

```python
# Load selected columns
df_excel_subset = pd.read_excel(
    io ="titanic.xlsx",
    usecols = ['Survived', 'Pclass', 'Age', 'Fare']
)
print(df_excel_subset.head())

rows, columns = df_excel_subset.shape
print(f"Rows: {rows} : Columns: {columns}")
```

**Check for sheet name in an excel file**

```python
# Check for sheet name in an excel file
xls = pd.ExcelFile("titanic.xlsx")
print(xls.sheet_names)
```

```python
# Load an excel file + sheet name
df_excel_sheet = pd.read_excel('titanic.xlsx', sheet_name='Sheet1')
```

**Load Parquet File**

```python
df_parquet = pd.read_parquet('titanic.parquet')

df_parquet.head()
```

```python
df_parquet_subset = pd.read_parquet(
    "titanic.parquet",
    columns = ['Survived', 'Pclass', 'Age', 'Fare']
)

rows, columns = df_parquet_subset.shape
print(f"Rows: {rows} : Columns: {columns}")
```
