```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

**Simple Line: y = 2x + 1**

```python
# Create a function
def lineFunc(x):
    return 2 * x + 1
```

```python
x = np.linspace(-3, 3, 100)
y = lineFunc(x)
```

```python
print(x[:5])
print(y[:5])
```

```python
# Find the slope of the x & y values
# Slope formula m = y2-y1/x2-x1

plt.figure(figsize=(6,5))
plt.plot(x, y, color="blue")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Linear Slope")
plt.grid()
```

### SLOPE
**Choose two points on the line**

```python
x1, x2 = -2, 1
y1, y2 = lineFunc(x1), lineFunc(x2)

print(x1, x2)
print(y1, y2)

# Find and calculate the slope
rise = y2 - y1
run = x2 - x1
slope = rise/run

print(f"Point A - Rise ({x2}, {x1})")
print(f"Point B - Run ({y2}, {y1})")
print(f"Slope: {slope}")
```

### Curve: y =2**2

```python
def curve(x):
    return x**2

x1 = np.linspace(-3, 3, 200)    
y1 = curve(x1)

plt.figure(figsize=(6,5))
plt.plot(x1, y1, color = "red")
plt.xlabel("x")
plt.ylabel("x**2")
plt.title("y=x**2")
plt.grid(alpha = 0.6)
plt.show()
```

**3 Estimating slope with a small change in x**

Idea: If we move x a tiny bit and see how much y changes, that change in y divided by change in x gives an approximate slope.

slope = dy/dx
- dy = change in y
- dx = change in x

```python
def slope_estimate(func, x1, h= 0.00001):
    x2 = x1 + h
    y1 = curve(x1)
    y2 = func(x2)
    dy = y2 - y1
    dx = x2 - x1
    slope = dy / dx
    return slope
```

```python
points = [-2, 5, 3]
print("Slope Estimate For Y**2 Using Small Change in X")

for p in points:
    s = slope_estimate(curve, p)
    print(f"x = {p:.1f} ==> Estimated Slope ==> {s:.2f}")
```

### CALCULUS PART 2

```python
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")
np.set_printoptions(suppress=True)
```

```python
x = np.array([1,2,3])
y_true = np.array([4,5,6])
```

### Updating a Weight
**w = w - lr * (dL/dw)**
- If dL/dw is positive, increasing w increases loss
- If dL/dw is negative, decreasing w increases loss, so we increase w

```python
# f(x) = w * x 

# y = w * x 

# Starting guess of w 
w = 4.0

# Learning rate
lr = 0.01

def loss(w):
    y_pred = w * x
    return np.mean((y_pred-y_true)**2)

# dloss => dL/dw
def dloss(w):
    y_pred = w * x
    return np.mean(2 * x * (y_pred - y_true))

print("Start w:", w, "loss", loss(w) )

for step in range(30):
    slope = dloss(w)
    w = w - lr * slope
    print("Step", step + 1, "w:", round(w, 3), "loss", round(loss(w), 3))
```
