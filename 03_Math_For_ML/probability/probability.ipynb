```python
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")
```

```python
np.random.seed(42)

n = 500
outcomes = np.random.choice(["H", "T"], size=n)

heads = np.sum(outcomes == "H")
tails = np.sum(outcomes == "T")

print(f"Heads: {heads}")
print(f"Tails: {tails}")
print(f"Estimated Outcome: P(H): {heads/n}")
print(f"Estimated Outcome: P(T): {tails/n}")
```
