```python
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")
```

**Coin Toss**

```python
np.random.seed(42)

n = 500
outcomes = np.random.choice(["H", "T"], size=n)

heads = np.sum(outcomes == "H")
tails = np.sum(outcomes == "T")

print(f"Heads: {heads}")
print(f"Tails: {tails}")
print(f"Estimated Outcome: P(H): {heads/n}")
print(f"Estimated Outcome: P(T): {tails/n}")
```

**Dice Roll**

```python
n = 1000

dice = np.random.randint(1, 7, size = n)

p_four = np.mean(dice == 4)
print(f"Estimated P(Four): {p_four}")
```

```python
number = np.random.randint(1, 1000, size = 40)

vals, counts = np.unique(dice, return_counts= True)
print(vals)
print(counts)

plt.figure(figsize=(6,5))
plt.bar(vals, counts)
plt.title("Dice Roll Outcomes")
plt.xlabel("values")
plt.ylabel("counts")
plt.show()
```
