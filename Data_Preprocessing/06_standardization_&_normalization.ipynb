Standardization and normalization are commonly used in **data preprocessing** to ensure that numerical features are on comparable scales. This can improve the performance of many machine learning algorithms, especially those that are sensitive to the scale of the input variables.

**NOTE:** The scaling parameters should be **fitted on the training data only** before being applied to the test data, to prevent data leakage.

Standardization and normalization are typically applied to **numerical features**, particularly when different variables have very different ranges or units.

There are two common approaches to scaling numerical data:

* **Standardization:** Transform the data so that it has a mean of 0 and a standard deviation of 1. This is useful when the features have different scales but the distribution of the data is reasonably well behaved. It is commonly performed using `StandardScaler`.

* **Normalization:** Rescale the data so that values fall within a specified range, commonly between 0 and 1. This is useful when you want features to have a consistent bounded range. It is commonly performed using `MinMaxScaler`.

**Goal**
- Apply StandardScalar or MinMaxScalar (only one is needed)
- Fit on train data
- Transofrm both fit and test data
