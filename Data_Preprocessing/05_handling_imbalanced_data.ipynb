Choosing not to balance our data before modelling can introduce bias into our predictions, especially when dealing with classification problems.

**NOTE:** `train_test_split` must be performed **before** sampling to prevent data leakage.

Class imbalance and dataset balancing are typically addressed in **classification problems**, where one class has significantly more observations than another.

There are three common approaches to handling imbalanced datasets:

* **Oversampling:** Increase the number of observations in the minority class to make it more comparable to the majority class.
* **Undersampling:** Reduce the number of observations in the majority class to make it more comparable to the minority class.
* **SMOTE (Synthetic Minority Over-sampling Technique):** Generate synthetic observations for the minority class to reduce class imbalance without simply duplicating existing observations.

```python
# %pip install imbalanced-learn
import matplotlib.pyplot as plt
import numpy as np  #ignore
import pandas as pd
import seaborn as sns
from imblearn.over_sampling import SMOTE, RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler
from sklearn.model_selection import train_test_split

# warnings.filterwarnings("ignore") 
pd.set_option("display.max_columns", None)
sns.set_style("whitegrid")

%matplotlib inline
```

**Load Data**

```python
DATA_PATH = r"C:\Users\Benny\Desktop\Machine-Learning-Bootcamp\Data_Preprocessing\diabetes.csv"

df = pd.read_csv(DATA_PATH)
print(f"Rows: {df.shape[0]}, Column: {df.shape[1]}")
```

```python
df.isna().sum()
```

```python
print("Outcome distribution (full data)")
print(df["Outcome"].value_counts())
print(df["Outcome"].value_counts(normalize=True).round(3).mul(100))
```

**Splitting The Dataset**

```python
X = df.drop(columns = ["Outcome"])
y = df["Outcome"]
print(f"X Shape: {X.shape}")
print(f"y Shape: {y.shape}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    test_size=.2,
    random_state=42,
    stratify=y
)
```

```python
print("Shapes after split:")
print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
print("Outcome distribution in TRAIN")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True).round(3).mul(100))
```

**Random Over Sampler(TRAIN Only)**

```python
ros = RandomOverSampler(random_state=42)
X_train_oversampled, y_train_oversampled = ros.fit_resample(X_train, y_train)

print(f"y_train_oversample {y_train_oversampled.value_counts()}")
print(f"y_train_oversampled_distribution {y_train_oversampled.value_counts(normalize=True).mul(100)}")
```

**Random Under Sampler**

```python
rus = RandomUnderSampler(random_state=42)
X_train_undersampled, y_train_undersampled = rus.fit_resample(X_train, y_train)

print(f"y_train_undersample {y_train_undersampled.value_counts()}")
print(f"y_train_undersampled_distribution {y_train_undersampled.value_counts(normalize=True).mul(100)}")
```

**SMOTE**

```python
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print(f"y_train_SMOTE: {y_train_smote.value_counts()}")
print(f"y_train_SMOTE_distribution: {y_train_smote.value_counts(normalize = True).mul(100)}")
```

**Save to excel**

```python
train_smote = X_train_smote.copy()
train_smote["Outcome"] = y_train_smote.values

test_original = X_test.copy()
test_original["Outcome"] = y_test.values

train_smote.to_csv("diabetes_train_balanced_smote.csv", index=False)
test_original.to_csv("diabetes_test_original.csv", index=False)
```
